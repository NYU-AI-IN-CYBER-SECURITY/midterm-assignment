```python
# RUN THIS AFTER YOU CLOSE UNSLOTH STUDIO! List all the models generated
import os
from google.colab import files

# Directly target the Unsloth Studio export directory
target_dir = "/root/.unsloth/studio/exports/"
file_paths = []

print(f"Scanning '{target_dir}' for exported models...")

# 1. Manually walk through the known directory structure
if os.path.exists(target_dir):
    for root, dirs, files_list in os.walk(target_dir):
        for file in files_list:
            if file.endswith(".gguf") and "ggml-vocab" not in file:
                full_path = os.path.join(root, file)
                file_paths.append(full_path)
else:
    print(f"⚠️ Warning: The directory '{target_dir}' was not found.")

# 2. Display selection menu if models are found
if not file_paths:
    print("❌ No GGUF files found inside the export directory. Check the path manually.")
else:
    print("\n--- Found the following GGUF models ---")
    for index, path in enumerate(file_paths, start=1):
        file_size = os.path.getsize(path) / (1024 * 1024 * 1024) # Convert to GB
        print(f"[{index}] {os.path.basename(path)} ({file_size:.2f} GB)")
        print(f"    Path: {path}\n")

    # 3. Handle user input choice
    try:
        choice = input(f"Enter the model number you want to download (1-{len(file_paths)}): ").strip()
        choice_idx = int(choice) - 1

        if 0 <= choice_idx < len(file_paths):
            selected_file = file_paths[choice_idx]
            print(f"\n🚀 Preparing to download: {os.path.basename(selected_file)}")
            print("⚠️ Keep this browser tab open and active during large file downloads.")

            # Trigger native browser download
            files.download(selected_file)
        else:
            print("❌ Invalid selection. Please enter a number listed above.")
    except ValueError:
        print("❌ Please enter a valid integer.")
    except Exception as e:
        print(f"❌ Download failed or cancelled: {e}")
```
